The purpose of this notebook was to confirm two things:
- Docling exposes section/header structure in a usable way.
- Docling gives page provenance for text blocks.

```python
from docling.document_converter  import DocumentConverter
from pathlib import Path
```

```python
PROJECT_ROOT = Path.cwd().parent
PAPER_DIR = PROJECT_ROOT / 'data' / 'papers'

sample_pdf = PAPER_DIR / '01_rag_lewis_2020.pdf'

converter = DocumentConverter()
result = converter.convert(sample_pdf)
```

```python
doc = result.document
```

```python
for item, level in doc.iterate_items():
    text = getattr(item, 'text', '')


    print(
        "LEVEL:", level,
        "| TYPE:", type(item).__name__,
        "| TEXT:", text[:120]
    )
```

```python
for item, level in doc.iterate_items():
    text = getattr(item, 'text','')

    if text:
        print('TEXT:' , text[:100])
        print('PROV:', getattr(item, 'prov', None))
        print('-'* 80)
```
