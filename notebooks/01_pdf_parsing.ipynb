```python
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent

PAPER_DIR = PROJECT_ROOT / 'data' / 'papers'
PROCESSED_DIR = PROJECT_ROOT /'data'/'processed'

PROCESSED_DIR.mkdir(parents=True, exist_ok=True)
```

```python
from docling.document_converter import DocumentConverter
```

```python
pdf_path = PAPER_DIR / '01_rag_lewis_2020.pdf'
convertor = DocumentConverter()

result = convertor.convert(pdf_path)
```

```python
doc = result.document
markdown = doc.export_to_markdown()

output_path = PROCESSED_DIR / "01_rag_lewis_2020.md"

output_path.write_text(
    markdown,
    encoding="utf-8"
)

print(output_path)
```

```python
for file in PAPER_DIR.glob('*.pdf'):
    result = convertor.convert(file)
    processed_doc = result.document
    markdown = processed_doc.export_to_markdown()

    output_path = PROCESSED_DIR / f'{file.stem}.md'
    output_path.write_text(
                            markdown,
                            encoding="utf-8"
                            )
    print(f"Processed: {file.name}")
```
